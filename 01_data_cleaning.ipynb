```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
ipl = pd.read_csv("E:\\ipl_dataset1\\IPL.csv",dtype = str)
```

```python
print("Matches shape:",ipl.shape)
print("\n--- MATHES COLUMNS ---")
print(ipl.columns.tolist())
```

```python
drop_cols = [
    'ball_no',          
    'bat_pos',          
    'non_striker_pos', 
    'valid_ball',      
    'runs_not_boundary',
    'umpire',           
    'gender',           
    'team_type',        
    'balls_per_over',
    'day',              
    'event_match_no',   
    'batter_runs',      
    'batter_balls',     
    'bowler_wicket',    
    'striker_out',      
]

ipl.drop(columns=drop_cols, inplace=True, errors='ignore')
print("Dropped columns")
print("Remaining columns:", len(ipl.columns))
```

```python
ipl['extra_type'].fillna('none', inplace=True)
ipl['wicket_kind'].fillna('none', inplace=True)
ipl['player_out'].fillna('none', inplace=True)
ipl['fielders'].fillna('none', inplace=True)
ipl['method'].fillna('Normal', inplace=True)
ipl['result_type'].fillna('Normal', inplace=True)
ipl['superover_winner'].fillna('none', inplace=True)
ipl['new_batter'].fillna('none', inplace=True)
ipl['next_batter'].fillna('none', inplace=True)

print("Nulls handled")
```

```python
ipl['date'] = pd.to_datetime(ipl['date'])
ipl['runs_batter'] = pd.to_numeric(ipl['runs_batter'], 
                                    errors='coerce')
ipl['runs_total'] = pd.to_numeric(ipl['runs_total'], 
                                   errors='coerce')
ipl['over'] = pd.to_numeric(ipl['over'], errors='coerce')

print("Data types fixed")
```

```python
team_map = {
    'Delhi Daredevils': 'Delhi Capitals',
    'Deccan Chargers': 'Sunrisers Hyderabad',
    'Kings XI Punjab': 'Punjab Kings',
    'Rising Pune Supergiant': 'Rising Pune Supergiants'
}

ipl['batting_team'] = ipl['batting_team'].replace(team_map)
ipl['bowling_team'] = ipl['bowling_team'].replace(team_map)
print("Team names standardized")
```

```python
ipl['phase'] = pd.cut(
    ipl['over'],
    bins=[-1, 6, 15, 20],  # changed 0 to -1
    labels=['Powerplay', 'Middle', 'Death']
)

print("Remaining phase nulls:", ipl['phase'].isnull().sum())
```

```python
ipl.to_csv("E:\\ipl_dataset1\\IPL.csv")
```

```python
print("Shape:", ipl.shape)
print("\nRemaining nulls:")
nulls = ipl.isnull().sum()
print(nulls[nulls > 0])

print("\nSeasons:", sorted(ipl['season'].unique()))
print("Total matches:", ipl['match_id'].nunique())
print("Total teams:", ipl['batting_team'].nunique())
```

```python
ipl['runs_target'].fillna(0, inplace=True)
ipl['win_outcome'].fillna('In Progress', inplace=True)
print(ipl[ipl['phase'].isnull()]['over'].unique())
```

```python
nulls = ipl.isnull().sum()
print(nulls[nulls > 0])
```

```python
ipl['batting_team'] = ipl['batting_team'].str.replace(
    'Royal Challengers Bangalore', 
    'Royal Challengers Bengaluru'
)
ipl['bowling_team'] = ipl['bowling_team'].str.replace(
    'Royal Challengers Bangalore', 
    'Royal Challengers Bengaluru'
)

rcb = ipl[ipl['batting_team'].str.contains('Royal', na=False)]['batting_team'].unique()
print(rcb)
```
